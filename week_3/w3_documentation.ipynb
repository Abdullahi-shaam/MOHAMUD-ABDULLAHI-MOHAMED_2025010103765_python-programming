Activity 1: Identify the Components

1.1 Inputs
1.1 Inputs
mark1 = First quiz mark entered by the user.        (Float) Data type 
mark2 = Second quiz mark entered by the user.       (Float) Data type
mark3 = Third quiz mark entered by the user.        (Float) Data type
another =User's answer to continue the program.     (yes/no) (String) Data type

1.2 What is the Process?
1. Receive three quiz marks from the user
2. Calculate the average: `average = (mark1 + mark2 + mark3) / 3`
3. Compare the average against the passing threshold (50)
4. Determine pass or fail based on the comparison
5. Ask if another student's data should be entered
6. Repeat or terminate based on user response

1.3 The output
Output	
Average Mark = The average of the three quiz marks
PASS =         Displayed when the average is 50 or above
FAIL=          Displayed when the average is below 50


Activity 2: Design the Algorithm
2.1 Pseudocode

BEGIN

REPEAT

 Display "Student Quiz Average Calculator"

 Input mark1

 Input mark2

 Input mark3

 average ← (mark1 + mark2 + mark3) / 3

 Display average

 IF average ≥ 50 THEN

  Display "PASS"

 ELSE

  Display "FAIL"

 END IF

 Input another

UNTIL another ≠ "yes"

Display "Program ended. Goodbye!"

END

2.2 Algorithm Structure
Sequence

Sequence refers to instructions that are executed one after another in order.

Examples:

Enter mark1
Enter mark2
Enter mark3
Calculate the average
Display the average
Selection

Selection is used when the program needs to make a decision.

Example:

if average >= 50:
    print("Result: PASS")
else:
    print("Result: FAIL")

The program checks the average mark and decides whether the student passes or fails.

Iteration

Iteration means repeating a set of instructions.

Example:

while True:

The loop allows the user to enter marks for multiple students. The program continues running until the user enters something other than "yes".
